### Setup — Import Libraries and Load Data

```python
import re
import string
import nltk
from nltk.tokenize import word_tokenize, TweetTokenizer

nltk.download('punkt', quiet=True)
nltk.download('punkt_tab', quiet=True)

DATA_PATH = r"D:\TXSA\Part A Dataset\Part A Dataset\Data_1.txt"

with open(DATA_PATH, "r", encoding="utf-8") as f:
    raw = f.read().strip()

text = re.sub(r'^\d+\t', '', raw, flags=re.MULTILINE)

print("Text corpus:")
print(text)
```

## Q5(a) — Implement Tokenization using NLTK TweetTokenizer
*(3 marks)*

**TweetTokenizer** is an NLTK tokenizer specifically designed to handle **informal, social-media-style text**. It extends standard tokenization with rules for:
- Preserving **hashtags** (`#NLP`) and **mentions** (`@user`) as single tokens
- Handling **emoticons** and **emoji** correctly
- Optionally **stripping handles**, **reducing repeated characters** (`looooong` → `loong`), and **lowercasing**

Three configurations are demonstrated below to show its flexibility.

```python
# ── Configuration 1: Default TweetTokenizer ────────────────────────────────
tknzr_default = TweetTokenizer()
tokens_tweet_default = tknzr_default.tokenize(text)

print(f"[TweetTokenizer — Default]  →  {len(tokens_tweet_default)} tokens")
print(tokens_tweet_default)
```

```python
# ── Configuration 2: Lowercase + Strip punctuation handles ────────────────
tknzr_lower = TweetTokenizer(preserve_case=False, strip_handles=True)
tokens_tweet_lower = tknzr_lower.tokenize(text)

print(f"[TweetTokenizer — Lowercase, strip handles]  →  {len(tokens_tweet_lower)} tokens")
print(tokens_tweet_lower)
```

```python
# ── Configuration 3: Reduce repeated characters ───────────────────────────
# Example with informal text to show repeated-char reduction
informal_text = "Classificationnnn is sooooo interesting!!! #NLP @researcher"
tknzr_reduce = TweetTokenizer(reduce_len=True, preserve_case=False)
tokens_informal = tknzr_reduce.tokenize(informal_text)

print(f"[TweetTokenizer on informal text — reduce_len=True]")
print(f"  Input  : {informal_text}")
print(f"  Output : {tokens_informal}")
print()

# Use default for the main corpus
tokens_tweet = tokens_tweet_default
print(f"Selected tokens for comparison (default config):")
print(tokens_tweet)
```

## Q5(c) — Why TweetTokenizer is Different / Better / Worse
*(5 marks)*

### How TweetTokenizer Works
TweetTokenizer uses a set of **ordered regular expressions** tailored for noisy, user-generated text. Its rules handle, in priority order:
1. **HTML entities** (`&amp;`, `&lt;`)
2. **URLs** (`http://...`)
3. **Twitter handles** (`@username`)
4. **Hashtags** (`#topic`)
5. **Emoticons** (ASCII `:-)` and Unicode 😊)
6. **Repeated punctuation** (`!!!` → `!`)
7. **Standard word/punctuation splitting**

---

### Where TweetTokenizer is **Better**

| Scenario | Why TweetTokenizer wins |
|----------|-------------------------|
| **Social media corpora** | Preserves `#hashtags` and `@mentions` as atomic tokens, which is critical for sentiment analysis on Twitter/Reddit data. NLTK splits these into `#` + `word`. |
| **Emoticon-aware sentiment** | Treats `:-)` and `😊` as single feature tokens rather than splitting them into meaningless characters. |
| **Noisy repeated characters** | `reduce_len=True` normalises `sooooo` → `sooo`, preventing vocabulary explosion in informal corpora. |
| **Built-in lowercasing** | `preserve_case=False` removes an extra preprocessing step, making pipelines more concise. |

### Where TweetTokenizer is **Worse** / Different

| Scenario | Why NLTK word_tokenize is better |
|----------|----------------------------------|
| **Formal academic text** | Data_1.txt contains no hashtags, mentions, or emoticons — TweetTokenizer provides no advantage and outputs nearly identical tokens. |
| **Sentence-boundary handling** | NLTK + `sent_tokenize` is more reliable for multi-sentence formal text; TweetTokenizer treats the entire string as one block. |
| **Penn Treebank compatibility** | NLTK's tokenizer follows Penn Treebank conventions, making it directly compatible with standard POS taggers and parsers. TweetTokenizer's output may not align with these tools. |

### Conclusion
For **Data_1.txt** (a formal academic passage), TweetTokenizer produces **results nearly identical to NLTK `word_tokenize`** since none of its special features (hashtags, emoticons, repeated characters) are triggered. The tokenizer's true strength emerges with **social media or user-generated content** — for example, in a sentiment analysis task on tweets, TweetTokenizer would significantly outperform NLTK by correctly capturing `#NLP`, `@researcher`, and `😊` as meaningful single tokens. In this assignment context, it is **different but not better** for formal text.
