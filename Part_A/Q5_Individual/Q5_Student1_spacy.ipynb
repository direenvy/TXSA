### Setup — Import Libraries and Load Data

```python
import re
import string
import nltk
import spacy
from nltk.tokenize import word_tokenize

nltk.download('punkt', quiet=True)
nltk.download('punkt_tab', quiet=True)

DATA_PATH = r"D:\TXSA\Part A Dataset\Part A Dataset\Data_1.txt"

with open(DATA_PATH, "r", encoding="utf-8") as f:
    raw = f.read().strip()

text = re.sub(r'^\d+\t', '', raw, flags=re.MULTILINE)

print("Text corpus:")
print(text)
```

---
## Q5(a) — Implement Tokenization using SpaCy
*(3 marks)*

**SpaCy** is an industrial-strength NLP library that uses a **rule-based tokenizer** combined with **exception dictionaries** (for contractions, abbreviations, etc.) and trained statistical models. Unlike NLTK, SpaCy processes text through a full NLP pipeline, producing rich `Token` objects with attributes such as POS tag, lemma, dependency, and more — all in a single pass.

```python
# Load SpaCy English model (small)
nlp = spacy.load("en_core_web_sm")

# Tokenise using SpaCy
doc = nlp(text)
tokens_spacy = [token.text for token in doc]

print(f"[SpaCy Tokenizer]  →  {len(tokens_spacy)} tokens")
print(tokens_spacy)
```

```python
# SpaCy provides rich token-level attributes in a single pipeline pass
print(f"{'Token':<20} {'Lemma':<20} {'POS':<8} {'Is Alpha':<10} {'Is Stop'}")
print("-" * 70)
for token in doc:
    print(f"{token.text:<20} {token.lemma_:<20} {token.pos_:<8} {str(token.is_alpha):<10} {token.is_stop}")
```

## Q5(c) — Why SpaCy is Different / Better / Worse
*(5 marks)*

### How SpaCy Tokenizes
SpaCy's tokenizer operates in **two stages**:
1. **Exception rules** — handles special cases like contractions (`don't` → `do` + `n't`), abbreviations (`U.S.`), and URLs before any splitting occurs.
2. **Prefix/Suffix/Infix rules** — applies regex patterns to split on punctuation at the start/end/middle of tokens.

This is the same philosophy as NLTK's `word_tokenize`, but SpaCy wraps it inside a full **processing pipeline** (`nlp(text)`) that simultaneously computes:
- Part-of-speech tags (`token.pos_`)
- Lemmas (`token.lemma_`)
- Dependency labels (`token.dep_`)
- Stop word flags (`token.is_stop`)
- Named entity recognition (`token.ent_type_`)

---

### Where SpaCy is **Better** than NLTK

| Scenario | Why SpaCy wins |
|----------|----------------|
| **Full NLP pipeline** | One call returns tokens + POS + lemmas + NER simultaneously, whereas NLTK requires separate function calls for each. |
| **Speed at scale** | SpaCy's core is written in Cython and processes text significantly faster than NLTK on large corpora. |
| **Production use** | SpaCy is designed for production — it has thread-safe pipelines, model versioning, and active maintenance. |
| **Lemmatisation** | `token.lemma_` gives the dictionary form (e.g., *barked* → *bark*) without a separate function call, unlike NLTK's `WordNetLemmatizer`. |

### Where SpaCy is **Worse** / Different

| Scenario | Why NLTK is sufficient |
|----------|------------------------|
| **Lightweight tasks** | SpaCy requires loading a trained model (`en_core_web_sm` ≈ 12 MB) even for simple tokenisation; NLTK needs no model for `word_tokenize`. |
| **Token-count parity** | For plain English text like Data_1.txt, both methods produce nearly identical tokens — SpaCy adds no measurable advantage on tokenisation alone. |
| **Educational transparency** | NLTK's tokenization logic is easier to inspect and explain; SpaCy abstracts away the pipeline internals. |

### Conclusion
For this specific assignment (tokenisation of a short academic text), the **token output is nearly identical** between SpaCy and NLTK. The real advantage of SpaCy is its **integrated pipeline** — it is the better choice when downstream tasks (POS tagging, lemmatisation, NER) are also required, which eliminates the need for multiple separate NLTK calls. For a standalone tokenisation task, NLTK `word_tokenize` is simpler and equally accurate.
