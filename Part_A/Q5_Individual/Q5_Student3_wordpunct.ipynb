### Setup — Import Libraries and Load Data

````python
import re
import string
import nltk
from nltk.tokenize import word_tokenize, WordPunctTokenizer

nltk.download('punkt', quiet=True)
nltk.download('punkt_tab', quiet=True)

DATA_PATH = r"D:\TXSA\Part A Dataset\Part A Dataset\Data_1.txt"

with open(DATA_PATH, "r", encoding="utf-8") as f:
    raw = f.read().strip()

text = re.sub(r'^\d+\t', '', raw, flags=re.MULTILINE)

print("Text corpus:")
print(text)
````

---
## Q5(a) — Implement Tokenization using WordPunctTokenizer
*(3 marks)*

**WordPunctTokenizer** is an NLTK tokenizer based on the simple regular expression:

```
\w+ | [^\w\s]+
```

This pattern splits text into two distinct categories:
- **`\w+`** — sequences of word characters (letters, digits, underscore)
- **`[^\w\s]+`** — sequences of non-word, non-space characters (i.e., punctuation runs)

Unlike `split()` which leaves punctuation attached to words, WordPunctTokenizer always isolates every punctuation character as a separate token.

````python
wpt = WordPunctTokenizer()
tokens_wpt = wpt.tokenize(text)

print(f"[WordPunctTokenizer]  →  {len(tokens_wpt)} tokens")
print(tokens_wpt)
````

````python
# Categorise tokens: words vs punctuation
word_tokens  = [t for t in tokens_wpt if re.match(r'^\w+$', t)]
punct_tokens = [t for t in tokens_wpt if not re.match(r'^\w+$', t)]

print(f"Word tokens   ({len(word_tokens)}) : {word_tokens}")
print(f"Punct tokens  ({len(punct_tokens)}) : {punct_tokens}")
````

````python
# Demonstrate how WordPunctTokenizer handles edge cases
edge_cases = [
    "don't stop",
    "U.S.A. is great",
    "He said: 'Hello!'",
    "open-class classification",
]

print("Edge Case Comparison: NLTK vs WordPunctTokenizer\n")
print(f"{'Input':<35} {'NLTK word_tokenize':<45} {'WordPunctTokenizer'}")
print("-" * 110)
for case in edge_cases:
    nltk_out = word_tokenize(case)
    wpt_out  = wpt.tokenize(case)
    print(f"{case:<35} {str(nltk_out):<45} {str(wpt_out)}")
````

## Q5(c) — Why WordPunctTokenizer is Different / Better / Worse
*(5 marks)*

### How WordPunctTokenizer Works
WordPunctTokenizer uses a **single regular expression** — `\w+ | [^\w\s]+` — applied via `re.findall()`. This is mathematically equivalent to the Regex approach from Q1 (`\b\w+\b`) with one key difference: **punctuation runs are also returned as tokens** rather than discarded. Every non-whitespace, non-word character sequence becomes its own token, without exception.

---

### Where WordPunctTokenizer is **Better**

| Scenario | Why WordPunctTokenizer wins |
|----------|-----------------------------|
| **Predictability** | The single-regex rule is completely deterministic — no context dependency. Every punctuation character is guaranteed to be a separate token. |
| **Speed** | A single compiled regex pass is faster than NLTK's multi-rule tokenizer pipeline, important when processing millions of documents. |
| **Punctuation as features** | For tasks like **authorship attribution** or **writing style analysis**, punctuation frequency is a key feature. WordPunctTokenizer makes it trivial to extract and count punctuation tokens. |
| **Multilingual text** | The `\w+` pattern respects Unicode word characters, making it more naturally applicable to multilingual corpora than NLTK's English-focused rules. |

### Where WordPunctTokenizer is **Worse** / Different

| Scenario | Why NLTK word_tokenize is better |
|----------|----------------------------------|
| **Contractions** | WordPunctTokenizer splits `don't` → `don` + `'` + `t`, while NLTK correctly produces `do` + `n't` following linguistic convention. The NLTK split is semantically more meaningful. |
| **Hyphenated compounds** | `open-class` → `open` + `-` + `class` in WordPunctTokenizer, losing the compound meaning. NLTK treats it as a single token. |
| **Abbreviations** | `U.S.A.` → `U` + `.` + `S` + `.` + `A` + `.` — each letter becomes a separate token, inflating the token count and misrepresenting the entity. |
| **NLP pipeline compatibility** | NLTK's output follows Penn Treebank standards, which are expected by most English POS taggers and parsers. WordPunctTokenizer's splits may break downstream tools. |

### Conclusion
WordPunctTokenizer is a **fast, simple, and consistent** alternative that is particularly suitable when:
- Punctuation needs to be explicitly retained as features (e.g., stylometry, authorship identification)
- Speed is a priority over linguistic precision
- The text is multilingual or non-standard

For the academic text in Data_1.txt, it produces a **slightly higher token count** than NLTK due to splitting hyphenated words and abbreviations. Overall, it is **different from NLTK** in its handling of complex tokens, and **worse** for standard English NLP tasks that require linguistically accurate tokenisation — but **better** when predictability, speed, or punctuation-as-features are the priority.
