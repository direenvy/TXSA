## Part A — Q1: Tokenization & Stop Word / Punctuation Filtering
**Dataset:** `Data_1.txt`

### Setup — Import Libraries and Load Data

```python
import re
import string
import nltk
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords

# Download required NLTK resources
nltk.download('punkt', quiet=True)
nltk.download('punkt_tab', quiet=True)
nltk.download('stopwords', quiet=True)

# Load Data_1.txt
DATA_PATH = r"D:\TXSA\Part A Dataset\Part A Dataset\Data_1.txt"

with open(DATA_PATH, "r", encoding="utf-8") as f:
    raw = f.read().strip()

# Remove the leading line number (e.g. '1\t') added by the file format
text = re.sub(r'^\d+\t', '', raw, flags=re.MULTILINE)

print("Original Text:")
print(text)
```

---
## Q1(a) — Word Tokenization using `split()`, Regular Expression, and NLTK
*(2 marks)*

Three different tokenization approaches are demonstrated below.

```python
# ── Method 1: Python built-in split() ──────────────────────────────────────
tokens_split = text.split()

print(f"[Method 1] Python split()  →  {len(tokens_split)} tokens")
print(tokens_split)
```

```python
# ── Method 2: Regular Expression tokenizer ─────────────────────────────────
# \b\w+\b matches any sequence of word characters (letters, digits, _)
tokens_regex = re.findall(r'\b\w+\b', text)

print(f"[Method 2] Regex (re.findall)  →  {len(tokens_regex)} tokens")
print(tokens_regex)
```

```python
# ── Method 3: NLTK word_tokenize ───────────────────────────────────────────
tokens_nltk = word_tokenize(text)

print(f"[Method 3] NLTK word_tokenize()  →  {len(tokens_nltk)} tokens")
print(tokens_nltk)
```

---
## Q1(b) — Justification of the Most Suitable Tokenization
*(3 marks)*

```python
# Side-by-side comparison
print(f"{'Method':<18} {'Token Count':<14} {'Retains Punctuation?':<22} {'Handles Abbreviations?'}")
print("-" * 75)

methods = {
    "split()": tokens_split,
    "Regex":   tokens_regex,
    "NLTK":    tokens_nltk,
}

for name, toks in methods.items():
    has_punct = any(t in string.punctuation for t in toks)
    # Contractions / abbreviations will appear with apostrophe
    has_apos  = any("'" in t for t in toks)
    print(f"{name:<18} {len(toks):<14} {str(has_punct):<22} {str(has_apos)}")
```

### Justification

**NLTK `word_tokenize()` is the most suitable method for text analytics**, for the following reasons:

| Criterion | `split()` | Regex `\w+` | NLTK |
|-----------|-----------|------------|------|
| Handles punctuation | ❌ Attaches to words (e.g. `"advance."`) | ❌ Silently drops all punctuation | ✅ Separates punctuation as own tokens |
| Token count accuracy | Lowest | Middle | Highest |
| Handles contractions | ❌ | ❌ | ✅ Splits `don't` → `do` + `n't` |
| NLP pipeline compatibility | Poor | Moderate | Excellent (Penn Treebank standard) |

- `split()` retains punctuation **attached to words** (e.g. `"advance."`, `"variants."`) which pollutes the vocabulary and causes mismatches during feature extraction.
- The **Regex** approach silently discards all punctuation, losing structural cues like semicolons that signal list boundaries in this corpus.
- **NLTK** correctly separates punctuation as independent tokens while preserving the word text, making it fully compatible with downstream NLP steps such as POS tagging, stemming, and sentiment analysis.

## Q1(c) — Stop Word and Punctuation Removal
*(2 marks)*

```python
# Use NLTK tokens (most suitable, as justified above)
stop_words  = set(stopwords.words('english'))
punctuation = set(string.punctuation)

# Lowercase all tokens first
tokens_lower = [t.lower() for t in tokens_nltk]

# Identify stop words found in the corpus
stop_words_found = [t for t in tokens_lower if t in stop_words]

# Filter: remove stop words AND punctuation
filtered_tokens = [
    t for t in tokens_lower
    if t not in stop_words and t not in punctuation
]

print(f"Total NLTK tokens         : {len(tokens_lower)}")
print(f"Stop words found ({len(stop_words_found)})      : {stop_words_found}")
print()
print(f"Tokens after filtering ({len(filtered_tokens)}):")
print(filtered_tokens)
```

## Q1(d) — Importance of Filtering Stop Words and Punctuation
*(3 marks)*

Stop words (e.g. *the*, *is*, *a*, *and*, *of*, *in*) are high-frequency function words that carry **little to no discriminative semantic content**. Punctuation marks serve grammatical roles in written text but add no meaning to bag-of-words or feature-based representations.

### Reasons filtering is important in text analytics:

**1. Dimensionality Reduction**  
Stop words typically account for 40–60% of word occurrences. Removing them drastically shrinks the feature space, reducing computation time and memory requirements for models such as TF-IDF vectorisers and classifiers.

**2. Improved Model Performance**  
Classifiers trained on high-frequency noise words may learn spurious correlations. Retaining only content-bearing terms (nouns, verbs, adjectives) focuses the model on semantically meaningful signals that differentiate document classes.

**3. Better Similarity Metrics**  
In document clustering or information retrieval, cosine similarity between documents is distorted by stop words that appear uniformly across all documents. Filtering yields more accurate inter-document comparisons.

**4. Language Model & Topic Model Quality**  
For n-gram models and topic models (LDA), removing stop words prevents trivial bigrams (e.g. *is the*, *of a*) from dominating probability estimates, surfacing topic-relevant patterns such as `classification–labels` or `sequence–inputs` instead.
