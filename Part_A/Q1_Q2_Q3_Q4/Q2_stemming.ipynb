### Setup — Import Libraries and Load Data

```python
import re
import nltk
from nltk.tokenize import word_tokenize
from nltk.stem import PorterStemmer, LancasterStemmer
from nltk.corpus import stopwords

nltk.download('punkt', quiet=True)
nltk.download('punkt_tab', quiet=True)
nltk.download('stopwords', quiet=True)

DATA_PATH = r"D:\TXSA\Part A Dataset\Part A Dataset\Data_1.txt"

with open(DATA_PATH, "r", encoding="utf-8") as f:
    raw = f.read().strip()

text = re.sub(r'^\d+\t', '', raw, flags=re.MULTILINE)

# Tokenise and keep only alphabetic, non-stop-word tokens for clean comparison
stop_words = set(stopwords.words('english'))
all_tokens = word_tokenize(text)
tokens = [t.lower() for t in all_tokens if t.isalpha() and t.lower() not in stop_words]

print("Original Text:")
print(text)
print(f"\nContent words after stop word removal ({len(tokens)}):")
print(tokens)
```

---
## Q2(a) — Importance of Stemming in Text Analytics
*(3 marks)*

**Stemming** is the process of reducing inflected or derived word forms to their common base (stem). For example, *classification*, *classifying*, and *classified* are all reduced to `classif`.

### Why stemming matters:

**1. Vocabulary Normalisation**  
Without stemming, *classify*, *classifies*, and *classified* are treated as three distinct features. Stemming merges them, reducing sparsity and ensuring documents discussing the same concept are recognised as similar.

**2. Improved Information Retrieval**  
A search query for *classify* should also surface documents containing *classification* or *classifying*. Stemming enables this by mapping all variants to a single index term.

**3. Reduced Feature Space**  
In high-dimensional text representations (TF-IDF, Bag-of-Words), stemming reduces the number of unique features, lowering memory usage and training time for ML models.

**4. Better Model Generalisation**  
Classifiers generalise better when trained on stems rather than full inflected forms, because they see more training evidence per stem than per individual word form — particularly helpful with small corpora.

## Q2(b) — Stemming Demonstration: Regex, Porter, Lancaster
*(3 marks)*

```python
# ── Method 1: Regular Expression Stemmer ───────────────────────────────────
def regex_stemmer(word):
    """Strip common English suffixes using regex substitutions."""
    word = re.sub(r'(tion|tions|ness|ment|ments|ing|ly|er|est|ed|s)$', '', word)
    return word if len(word) > 2 else word

regex_stems = [regex_stemmer(t) for t in tokens]

print("[Method 1] Regular Expression Stemmer")
print(f"{'Original':<25} -> {'Stem'}")
print("-" * 40)
for orig, stem in zip(tokens, regex_stems):
    tag = "  ← changed" if orig != stem else ""
    print(f"{orig:<25} -> {stem}{tag}")
```

```python
# ── Method 2: Porter Stemmer ───────────────────────────────────────────────
porter = PorterStemmer()
porter_stems = [porter.stem(t) for t in tokens]

print("[Method 2] Porter Stemmer")
print(f"{'Original':<25} -> {'Stem'}")
print("-" * 40)
for orig, stem in zip(tokens, porter_stems):
    tag = "  ← changed" if orig != stem else ""
    print(f"{orig:<25} -> {stem}{tag}")
```

```python
# ── Method 3: Lancaster Stemmer ────────────────────────────────────────────
lancaster = LancasterStemmer()
lancaster_stems = [lancaster.stem(t) for t in tokens]

print("[Method 3] Lancaster Stemmer")
print(f"{'Original':<25} -> {'Stem'}")
print("-" * 40)
for orig, stem in zip(tokens, lancaster_stems):
    tag = "  ← changed" if orig != stem else ""
    print(f"{orig:<25} -> {stem}{tag}")
```

## Q2(c) — Comparison and Differences among Stemmers
*(4 marks)*

```python
# Full side-by-side comparison table
print(f"{'Original':<25} {'Regex':<20} {'Porter':<20} {'Lancaster'}")
print("-" * 85)
for orig, r_s, p_s, l_s in zip(tokens, regex_stems, porter_stems, lancaster_stems):
    print(f"{orig:<25} {r_s:<20} {p_s:<20} {l_s}")

# Count how many words each method changed
regex_changed    = sum(1 for o, s in zip(tokens, regex_stems)    if o != s)
porter_changed   = sum(1 for o, s in zip(tokens, porter_stems)   if o != s)
lancaster_changed = sum(1 for o, s in zip(tokens, lancaster_stems) if o != s)

print(f"\nWords changed  →  Regex: {regex_changed}  |  Porter: {porter_changed}  |  Lancaster: {lancaster_changed}")

# Average stem length
avg_regex     = sum(len(s) for s in regex_stems)    / len(regex_stems)
avg_porter    = sum(len(s) for s in porter_stems)   / len(porter_stems)
avg_lancaster = sum(len(s) for s in lancaster_stems) / len(lancaster_stems)
avg_orig      = sum(len(t) for t in tokens)          / len(tokens)

print(f"Avg stem length  →  Original: {avg_orig:.2f}  |  Regex: {avg_regex:.2f}  |  Porter: {avg_porter:.2f}  |  Lancaster: {avg_lancaster:.2f}")
```

### Explanation of Differences

| Feature | Regex Stemmer | Porter Stemmer | Lancaster Stemmer |
|---------|--------------|---------------|------------------|
| **Algorithm** | Manual suffix list | 5-phase rule cascade | Iterative rule application |
| **Aggressiveness** | Shallow | Moderate | Most aggressive |
| **Stem readability** | Moderate | Moderate | Often unreadable |
| **Over-stemming risk** | Low | Low–Medium | High |
| **Under-stemming risk** | High | Low | Very low |
| **Speed** | Fastest | Fast | Fast |

**1. Regular Expression Stemmer:**  
Rule-based — strips a fixed list of common suffixes (`ing`, `tion`, `ed`, `ly`, etc.) using regex substitution. It is the fastest and most transparent but very shallow. It misses irregular forms and has no linguistic awareness. For example, `multiclass` is unchanged (suffix not in the list), while `classification` → `classif` (correct direction but crude).

**2. Porter Stemmer (M.F. Porter, 1980):**  
Applies five sequential reduction phases with condition-based rules (measuring the number of vowel-consonant sequences in the stem). More conservative than Lancaster — prefers slightly longer, more readable stems. Well-balanced between over- and under-stemming. Example: `classification` → `classif`, `jointly` → `jointli`. The gold standard for most English NLP tasks.

**3. Lancaster Stemmer (Paice/Husk, 1990):**  
Highly aggressive iterative algorithm — applies rules repeatedly until no rule matches, producing the shortest possible stems. Maximally reduces the vocabulary but at the cost of readability and precision. Example: `classification` → `class`, `interesting` → `interest`, `assigned` → `assign`. High risk of over-stemming (merging semantically distinct words onto the same stem).

**Conclusion:** For this corpus, **Porter Stemmer** is the most suitable — it strikes the best balance between reducing vocabulary size and preserving enough stem information for accurate downstream analysis.
